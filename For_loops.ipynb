```python
#for loop
fruits=["apple", "banana", "cherry"]
for x in fruits:
    print(x)
        
```

```python
#looping through words
for x in "banana":
    print(x)    
```

```python
#break statement within for loop
fruits=["apple", "banana", "cherry"]
for x in fruits:
    print(x)
    if x == "banana":
        break
```

```python
#break statement within for loop
fruits=["apple", "banana", "cherry"]
for x in fruits:
    if x == "banana":
        break
    print(x)
```

```python
#continue statement within for loop
fruits=["apple", "banana", "cherry"]
for x in fruits:
    if x == "banana":
        continue
    print(x)
```

```python
#range function
for x in range(6):
    print(x)
```

```python
#nested for loops
adjectives=["red", "big", "tastly"]
fruits=["apple","banana", "cherry"]
for x in adjectives:
    for y in fruits:
        print(x,y)
        
```
