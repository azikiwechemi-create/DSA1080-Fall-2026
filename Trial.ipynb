```python
def calculate_grade(marks):
    if marks >= 70:
        return "A"
    elif marks >= 60:
        return "B"
    elif marks >= 50:
        return "C"
    elif marks >= 40:
        return "D"
    else:
        return  "F"

marks = float(input("Enter marks here", ))
grade = calculate_grade(marks)
print("Grade:", grade)
```
