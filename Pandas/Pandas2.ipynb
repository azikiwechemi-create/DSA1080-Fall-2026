```python
#Importing the csv file and pandas
import pandas as pd
df = pd.read_csv("DSA1080_Pandas_Student_Data.csv")
print(df.head())
print(df.shape)

#inspect the data
print(df.info())
print(df.describe())
print(df.isna().sum())
```

```python
#selecting columns
print(df["Name"])
print(df[["Name", "Score"]])
print(df["Score"].mean)
```

```python
#Selecting rows using loc
print(df.loc[0])
print(df.loc[0, "Name"])
print(df.loc[0, ["Name", "Score"]])
```

```python
#Selecting rows with iloc
print(df.iloc[0:5])
print(df.iloc[0:4], ["Name", "Score"])
```

```python
#filter table with a condition
high=df[df["Score"] >= 70]
print(high[["Name", "Score"]])
city=df[df["City"] == "Nairobi"]
print(city.head())
```

```python
#Combine conditions with & and|
selected = df[(df["Score"] >= 70) & (df["StudyHours"] >= 10)]
print(selected)
```

```python
#sort results to find the top perfomers
best = df.sort_values(by="Score", ascending=False)
print(best[["Name", "Score"]].head(5))
```

```python
#Understanding missing values
df=pd.read_csv("DSA1080_Pandas_Student_Data.csv")
print(df.isna().sum())
print(df[df["Score"].isna()])
```

```python
#Two ways of handling missing values
cleaned = df.copy()

median_score = cleaned["Score"].median()
cleaned["Score"] = cleaned["Score"].fillna(median_score)

cleaned["City"] = cleaned["City"].fillna("Unknown")
print(cleaned.isna().sum())
```

```python
#Remove duplicates and check data quality
print(cleaned.duplicated().sum())
cleaned = cleaned.drop_duplicates()

print(cleaned["Score"].between(0, 100).all())
print(cleaned.shape)
```

```python
#Summary Statistics
print(cleaned["Score"].mean())
print(cleaned["Score"].median())
print(cleaned["Score"].std())
print(cleaned["Score"].min())
print(cleaned["Score"].max())
print(cleaned["Score"].count())
```

```python
#Frequency counts and categories
print(cleaned["Course"].value_counts())
print(cleaned["Course"].value_counts())

print(cleaned["Course"].nunique())
```

```python
#Group data and compare courses
course_summary = cleaned.groupby("Course")["Score"].mean()
print(course_summary.sort_values(ascending=False))
```
