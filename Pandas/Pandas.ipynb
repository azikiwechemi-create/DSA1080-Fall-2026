```python
import pandas as pd
print(pd.__version__)
marks = [72,85,61,90]
scores = pd.Series(marks)
print(scores)
```

```python
scores= pd.Series([72,85,61,90], index=['Maths','Science','English','History'])
print(scores)
print(scores['Maths'])
```

```python
scores = pd.Series([72,85,61,90], index=['Maths','Science','English','History'])
print(scores.mean())
print(scores.max())
print(scores.min())
print(scores>70)
print((scores>70).sum())
```
