```python
# A number is even if the remainder from dividing by 2 == 0 
# If it is equal to 1 then the number is odd
Input_number= int(input("Enter a number", ))
if Input_number % 2 == 0:
    print("{} is Even" .format(Input_number))
else:
    print("{} is Odd" .format(Input_number))
    
```

```python
#Program to swap two variables
P = int(input("Please enter a value for P", ))
Q = int(input("Please enter a value for Q", ))
print("The value of P before swapping is", P)
print("The value of Q before swapping is", Q)
temp = P
P=Q
Q=temp
print("The value of P after swaping is:", P)
print("The value of Q after swaping is:", Q)

```

```python
#Finding the factorial of a given number
num = int(input("Enter a value", ))
factorial = 1
if num < 0:
    print("There are no negative values for factorial", )
elif num == 0:
    print("The factorial of 0 is 1")
else:
    for i in range(1,num + 1):
        factorial = factorial * i
        print("The factorial of", num ,"is" ,factorial)
```

```python
#finding the factorial of a given number
num = int(input("Enter a value", ))
factorial = 1
if num < 0:
    print("There is no factorial for negative numbers")
elif num == 0:
    print("The factorial for 0 is 1")
else:
    for i in range(1,num + 1):
        factorial = factorial*i
    print("The factorial for", num, "is", factorial)
```

```python
#finding the sum of natural numbers
num = int(input("Enter a number", ))
if num < 0:
    print("Enter a positive number")
else:
    sum = 0
    while(num > 0):
        sum += num
        num -= 1
    print("The sum is", sum)
```
