```python
#if statement
a=33
b=200
if b > a:
  print("b is greater than a")
  
```

```python
#elif statement
a=33
b=33
if b > a:
    print("b is greater than a")
elif a == b:
    print("a and b are equal")
```

```python
#else statement
a=34
b=33
if b > a:
    print("b is greater than a")
elif a == b:
    print("a and b are equal")
else:
    print("a is greater than b")
```

```python
#Short hand if statement
a = 200
b = 33
if a > b: print("a is greater than b")
```

```python
#short hand if else
a=200
b=33
print("A") if a > b else print("B")
```

```python
#multiple else statements
a = 200
b = 33
print("A") if a > b else print("=") if a == b else print("B")
```

```python
#And Statement
a = 200
b = 33
c = 500
if a > b and c > a:
    print("Both conditions are True")
```

```python
#OR statement
a = 200
b = 33 
c =500
if a > b or c > a:
    print("At least on of the conditions is True")
```

```python
#nested if statement
x=20
if x > 10:
    print("Above ten,")
    if x > 20:
        print("and also above 20!") 
    else:
        print("but not above 20.")
```

```python
#Check if a number is positive
num = float(input("Enter a number: "))
if num >= 0:
    if num == 0:
        print("Zero")
    else: 
        print("Positive number")
else: 
    print("Negative number")
```

```python
#while loop
i=1
while i <10:
    print(i)
    i += 1
```

```python
#while loop with break statement
i=1
while i < 10:
    print(i)
    if i==5:
        break
    i+=1
```
