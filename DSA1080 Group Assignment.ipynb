```python
#Part A

player= {"name": "Lamine", "age": 19, "position": "Forward", "team": "FC Barcelona", "nationality": "Spain", "goals": 15, "assists": 7, "foot": "left"}
type(player)
print(player)
```

```python
#Part B
print(player["name"])
print(player["age"])
print(player["position"])
print(player["team"])
print(player["nationality"])

print(player.get("goals"))
print(player.get("assists"))
print(player.get("foot"))

print(player.keys())
print(player.values())
print(player.items())
```

Keys act as unique immutable identifires used to rapidly look up data while values are the actual arbitrary data
stored inside. Every keys must be hashable (such as strings integers and tuples) and strictly unique. 

Values by contrast have no such restrictions they can be of any data type - mutable or immutable- including lists, dictionaries, and objects. They can freely duplicate across multiple different keys without conflict. 

```python
#Part C
player.update({"language": "Spanish"}) 
player.update({"height": 1.75,})
player.update({"weight": 70})
player.update({"siblings": 2})
print(player)
```

```python
#Part D
print('Height before update:', player['height'])
player['height'] = 1.80
print('Height after update:', player['height'])

print('Weight before update:', player['weight'])
player['weight'] = 75
print('Weight after update:', player['weight'])

print('Siblings before update:', player['siblings'])
player['siblings'] = 3
print('Siblings after update:', player['siblings'])

print('Language before update:', player['language'])
player['language'] = "English"
print('Language after update:', player['language'])


```

Height changed from 1.75 to 1.8

Siblings changed from 2 to 3
Language changed from Spanish to English

```python
#Part E
player_copy=player.copy()
player.pop("age")
print(player)
player.pop("position")
print(player)
player.__delitem__("team")
print(player)
player.clear()
print(player)
```

```python
#Part F

print(player_copy.keys())
print(player_copy.values())
print(player_copy.items())
print(player_copy.get("name"))


new_info = {"income": 50000, "sponsor": "Nike", "awards":"Golden Boot"}
player_copy.update(new_info)
print(player_copy)

player_backup = player_copy.copy()
print(player_backup)
```

1. Keys() gives us a view of all labels in dictionary
2. Values() gives us view of all data stored in dictionary
3. Items() gives us all the key value pairs together, each as a tuple
4. get() returns the value for a key 
5. update() merges another dictionary into the exsisting one
6. copy() creates a separate duplicate of the dictionary
