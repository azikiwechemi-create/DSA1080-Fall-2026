```python
#companies in the IT industry
it_companies = ["Facebook", "Google", "Microsoft", "Apple", "IBM", "Oracle", "Amazon"]
print("The IT companies are:", it_companies)

print("The first company is:", it_companies[0])
print("The middle company is:", it_companies[3])
print("The last company is:", it_companies[-1])


it_companies[4] = "Intel"
print("The IT companies after replacing IBM with Intel is:", it_companies)

it_companies.insert(3, "Sony")
print("The IT companies after adding Sony is:", it_companies)

#Change one of the it_companies names to uppercase
it_companies[1] = it_companies[1].upper()
print("Google in uppercase is:", it_companies[1])


#check if a certain company exists in the it_companies list
company_to_check = "Amazon"
if company_to_check in it_companies:
    print(f"{company_to_check} exists in the list.")
else:
    print(f"{company_to_check} does not exist in the list.")

#sort the it_companies list in ascending order
it_companies.sort()
print("The IT companies in ascending order are:", it_companies)

it_companies.sort(reverse=True)
print("The IT companies in descending order are:", it_companies)

#Slice out the first 3 companies from the list
first_three_companies = it_companies[:3]
print("The first three companies are:", first_three_companies)

#Slice out the last 3 companies from the list
last_three_companies = it_companies[-3:]
print("The last three companies are:", last_three_companies)

#Slice out the middle company from the list
middle_company = it_companies[3]
print("The middle company is:", middle_company)

#Remove the first IT company from the list
del it_companies[0]
print(it_companies)

#Remove the middle IT company from the list
del it_companies[3]
print(it_companies)

#Remove the last IT company from the list
del it_companies[-1]
print(it_companies)

#Remove all IT companies from the list
it_companies.clear()
print(it_companies)

#Destroy all IT companies from the list
del it_companies


```

```python
#sort the list
ages = [19, 22, 19, 24, 20, 25, 26, 24, 25, 24]

min_ages = min(ages)
max_ages = max(ages)

print("The minimum age in this list is:", min_ages)
print("The maximum age in this list is:", max_ages)

#Find the median ages
import statistics
median_age = statistics.median(ages)
print("The median age is:", median_age)

#Find the average age
average_age = sum(ages) / len(ages)
print("The average age is:", average_age)

#Find the range
range_age = max_ages - min_ages
print("The range for the ages is:", range_age)

min_diff = abs(min_ages - average_age)
max_diff = abs(max_ages - average_age)

print("Distance from min to average:", round(min_diff, 2))  # 3.8
print("Distance from max to average:", round(max_diff, 2))  # 3.2

# Compare the two values
print("Is min further from average than max?", min_diff > max_diff)  # True

```
