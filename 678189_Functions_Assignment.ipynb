```python
#Student Grade Calculator
def calculate_grade(marks):
    if marks >= 70:
        return "A"
    elif marks >= 60:
        return "B"
    elif marks >= 50:
        return "C"
    elif marks >= 40:
        return "D"
    else:
        return "F"

marks = float(input("Enter exam marks: "))
grade = calculate_grade(marks)
print("Grade:", grade)
```

```python
# Testing calculate_grade with different values
for m in [85, 70, 69, 60, 55, 45, 40, 39, 10]:
    print(m, "->", calculate_grade(m))
```

```python
#Shopping Bill calculator
def calculate_total(price, quantity):
    return price * quantity

price = float(input("Enter price per item (KES): "))
quantity = int(input("Enter quantity: "))
total = calculate_total(price, quantity)
print(f"Total cost: KES {total:,.2f}")
```

```python
# Testing calculate_total with different values
print(f"KES {calculate_total(120, 4):,.2f}")      # 480
print(f"KES {calculate_total(2500, 5):,.2f}")     # 12,500
print(f"KES {calculate_total(49.50, 3):,.2f}")    # 148.50
print(f"KES {calculate_total(100, 0):,.2f}")      # 0 items
```

```python
#Even Odd number checker
def check_number(number):
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"

number = int(input("Enter a number: "))
result = check_number(number)
print(f"{number} is {result}")
```

```python
# Testing check_number with even, odd, zero and a negative number
for n in [8, 7, 0, -3, 100]:
    print(n, "is", check_number(n))
```

```python
#Simple Interest Calculator
def simple_interest(principal, rate, time):
    return (principal * rate * time) / 100

principal = float(input("Enter principal amount (KES): "))
rate = float(input("Enter annual interest rate (%): "))
time = float(input("Enter time in years: "))
interest = simple_interest(principal, rate, time)
print(f"Simple interest: KES {interest:,.2f}")
```

```python
# Testing simple_interest with different values
print(f"KES {simple_interest(10000, 5, 2):,.2f}")    # 1,000
print(f"KES {simple_interest(50000, 8, 3):,.2f}")    # 12,000
print(f"KES {simple_interest(2500, 12, 0.5):,.2f}")  # 150
```

```python
#Student Performance report
def student_report(name, cat_marks, exam_marks):
    final_score = (cat_marks * 0.40) + (exam_marks * 0.60)
    if final_score >= 50:
        status = "Pass"
    else:
        status = "Fail"
    print("----- Student Report -----")
    print("Name:", name)
    print(f"Final Score: {final_score:.1f}")
    print("Status:", status)

name = input("Enter student name: ")
cat_marks = float(input("Enter CAT marks (out of 100): "))
exam_marks = float(input("Enter exam marks (out of 100): "))
student_report(name, cat_marks, exam_marks)
```

```python
# Testing student_report: a pass, a fail, and the 50% boundary
student_report("Amina", 60, 70)   # Pass
print()
student_report("Brian", 30, 40)   # Fail
print()
student_report("Cynthia", 50, 50) # Exactly 50, Pass
```
