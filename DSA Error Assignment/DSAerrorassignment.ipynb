##Section A
1. The first error that one could encounter is a value error when a user types in a string and in the code it is an int(input())
2. The second error that we could encounter is a zerodivision error where a user inputs zero 
3. The third error is a name error where a user does not input anything in the code meaning that the variable was never created


Exception handling matters because users make mistakes when inputing data and so handling those errors is what enables a system to be funcitonal. Because you don't want the system to crash and lose data. 

```python
#Section B

try:
    a = int(input("Enter a number: "))
except ValueError:
    print("Invalid input please enter a number.")
else:
    print("You age is:", a)
```

```python
#Safe marks calculator
#Safe marks calculator
try:
    total_marks = int(input("Enter your total marks: "))
    no_of_subjects = int(input("Enter the number of subjects: "))
    average_marks = total_marks / no_of_subjects
except ValueError:
    print("Invalid input. Please enter numbers only.")
except ZeroDivisionError:
    print("Invalid input. Number of subjects cannot be zero.")
else:
    print("Your average marks are:", average_marks)
finally:
    print("Calculation attempt completed.")
```

```python
#Students results file reader
try:
    with open("student_results.txt", "r") as file:
        results = file.readlines()
        for result in results:
            print(result.strip())
except FileNotFoundError:
    print("File not found.")
except Exception as e:
    print("An error occurred:", e)
```
