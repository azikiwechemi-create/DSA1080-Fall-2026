```python
# 1. Unpack siblings and parents from family_members
*siblings, father, mother = family_members
print("Siblings:", siblings)
print("Parents:", father, "and", mother)

# 2. Create fruits, vegetables, and animal_products tuples and join them
fruits = ("apple", "banana", "mango")
vegetables = ("carrot", "spinach", "onion")
animal_products = ("milk", "eggs", "meat")

food_stuff_tp = fruits + vegetables + animal_products
print("Food stuff tuple:", food_stuff_tp)

# 3. Change food_stuff_tp tuple to food_stuff_lt list
food_stuff_lt = list(food_stuff_tp)
print("Food stuff list:", food_stuff_lt)

# 4. Slice out the middle item or items
n = len(food_stuff_lt)
mid = n // 2
if n % 2 == 1:
    middle_items = food_stuff_lt[mid : mid + 1]
else:
    middle_items = food_stuff_lt[mid - 1 : mid + 1]
print("Middle item(s):", middle_items)

# 5. Slice out the first three and last three items from food_stuff_lt
first_three = food_stuff_lt[:3]
last_three = food_stuff_lt[-3:]
print("First three:", first_three)
print("Last three:", last_three)

# 6. Delete the food_stuff_tp tuple completely
del food_stuff_tp

# 7. Check if an item exists in nordic_countries
nordic_countries = ('Denmark', 'Finland', 'Iceland', 'Norway', 'Sweden')
print("Is Estonia a Nordic country?", "Estonia" in nordic_countries)
print("Is Iceland a Nordic country?", "Iceland" in nordic_countries)
```
