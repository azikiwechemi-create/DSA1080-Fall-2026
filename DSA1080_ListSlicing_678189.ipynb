```python
#Part A Creating the list
weekly_steps = [7824, 10333, 7445, 10852, 6386, 9315, 5785, 6674, 7689, 11689, 10518, 8034, 6559, 12007]
print(weekly_steps)
len(weekly_steps)
```

```python
#Part B Basic Indexing

print("First value:", weekly_steps[0])
print("Fourth value:", weekly_steps[3])
print("Last value:", weekly_steps[-1])
print("Second to last value:", weekly_steps[-2])
```

An Index usually tells python how far a value is from the start of the list. So if the first item is indexed as 0 then the next item is 1 step away.


```python
#Part C Basic List slicing

first_five = weekly_steps[0:5]
print("First five values:", first_five)
middle_days = weekly_steps[5:10]
print("Middle five values:", middle_days)
last_four = weekly_steps[-4:]
print("Last four values:", last_four)
inner_days = weekly_steps[1:-1]
print("Inner values:", inner_days)
```

In general, python start its index with at the start list[start:stop] and goes up to stop but does not include the stop index but the value before the stop index. 

For example [0:5] let's python know to start at index 0 and go all the way to index 5 and then exclude index 5. 

```python
#Part D Slicing with steps

alternate_days = weekly_steps[0:14:2]
print("Alternate days:", alternate_days)
alternate_from_day2 = weekly_steps[1:14:2]
print("Alternate days starting from day 2:", alternate_from_day2)
reversed_steps = weekly_steps[::-1]
print("Reversed list:", reversed_steps)
reverse_alternate = weekly_steps[-1:0:-2]
print("Reverse alternate days:", reverse_alternate)
```

[::2] Start and stop here are blank so Python defaults to the beginning and the end of a list. The 2 means it takes an item and skips one and then takes the next one.

[1::2] This one says that Python should start at Index 1 and go all the way to the end. By starting from index 1 it will skip an item and then take the next one.

[::-1] The start and stop are blank and because the step is negative python will start at the end of the list making its way to the start of the list by taking a step of 1 

[::-2] The start and stop are blank and because the step is negative python will start at the end of the list making its way to the start of the list by taking steps of 2
