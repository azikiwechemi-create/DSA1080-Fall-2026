1. A string is any data type that is input inside "" marks. It is mostly used to contain text in Python
2. A list is an ordered an mutable collection of items in python that allows duplicate values
3. A tuple is an ordered immutable collection of items meaning that the contents cannot be added, changed or removed.
4. A set is an unordered, unique, and immutable collections of items in Python. The set itself is mutable meaning that items can be added or removed though the elements inside must be immutable.



```python
sentence = "I am a teacher and I love to inspire and teach people"

words = sentence.split()
unique_words = set(words)

print("Unique words:", unique_words)
print("Number of unique words:", len(unique_words))  # 10
```
